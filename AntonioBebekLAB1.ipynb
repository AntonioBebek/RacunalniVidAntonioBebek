```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("lab1/images/pepper.bmp")

hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

hsv[:, :, 0] = (hsv[:, :, 0] + 30) % 180

rgb = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

plt.imshow(rgb)
plt.axis("off")
plt.show()
```
